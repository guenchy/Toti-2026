````python
notas = np.array([8.5, 5.0, 9.2])
````

````python
print(notas)
````

````python
print(notas)
print(notas * 2)
print(notas + 1)
print(notas / 10)
````

> Na vetorização, seus operadores são aplicados a todos os elementos

````python
# Trabalhando com duas arrays
provas = np.array([8.0, 6.0, 9.0])
trabalhos = np.array([9.0, 7.0, 10.0])

print(provas + trabalhos)
print((provas + trabalhos) / 2)
````

### É muito mais rápido!

````python
import time

n = 1_000_000
lista = list(range(n))
array = np.arange(n)

inicio = time.time()
dobro_lista = []
for x in lista:
    dobro_lista.append(x * 2)
tempo_lista = time.time() - inicio

inicio = time.time()
dobro_array = array * 2
tempo_array = time.time() - inicio

print(f"lista: {tempo_lista:.4f} segundos")
print(f"array: {tempo_array:.4f} segundos")
print(f"o array foi {tempo_lista / tempo_array:.0f}x mais rápido")
````

## 1.7 - Agregações

````python
notas = np.array([8.0, 5.5, 9.2, 4.1, 7.0])

print("soma: ", notas.sum())
print("média:", notas.mean())
print("maior:", notas.max())
print("menor:", notas.min())
print("desvio:", notas.std())
````
