## Como achar alunos do Rio de Janeiro que foram aprovados?

````python
df_aprovados = df[df["nota_final"] >= 7]
df_aprovados[df_aprovados["estado"] == "Rio de Janeiro"]
````

````python
df[(df["nota_final"] >= 7) and (df["estado"] == "Rio de Janeiro")]
````

>`and` compara um valor com outro. Se passarmos uma `Series` ele quebra. Pra isso temos o `&`

````python
rj_aprovados = df[(df["nota_final"] >= 7) & (df["estado"] == "Rio de Janeiro")]

print(len(rj_aprovados), "aprovados no Rio de Janeiro")
rj_aprovados.head()
````

**A regra, escrita na tela:**

| Um valor | Coluna inteira |
|---|---|
| `and` | `&` |
| `or` | `\|` |
| `not` | `~` |

>⚠️ **Os parênteses são obrigatórios.** Sem eles, o `&` tem prioridade sobre o `>=` e o
Python tenta comparar coisas erradas.


````python
# 🔮 Preveja: quantos alunos têm nota >= 7 OU são de São Paulo?
print(
    len(
        df[(df["nota_final"] >= 7) | (df["estado"] == "São Paulo")]
    )
)
````

## 3.2 Achado: alunos com `idade == 0`

````python
# 🎤 Investigar antes de decidir
print("alunos com idade 0:", (df["idade"] == 0).sum())
df[df["idade"] == 0][["nome", "ano_ingresso", "idade", "estado"]]
````

````python
df["idade"].value_counts()
````

> É um erro muito comum em bases reais: a pessoa não preenche a informação e o sistema coloca 0 ao invés de vazio

>⚠️ **ATENÇÃO**: zero é um número válido, então entra nas contas.

````python
# Olhe o que acontece com a nossa média de idade!
print("com os zeros: ", round(df["idade"].mean(), 2))
print("sem os zeros: ", round(df[df["idade"] > 0]["idade"].mean(), 2))
````

# E agora? Precisamos tomar uma decisão!

#### Nossas opções:


| Caminho | Vantagem | Custo |
|---|---|---|
| **Apagar as linhas** | simples, base fica limpa | perde as outras 8 informações dessas 3 pessoas |
| **Trocar 0 por `NaN`** | honesto: "não sabemos" | reduz o `n` das análises de idade |
| **Estimar pela média** | mantém o `n` | inventa dado; distorce a distribuição |

Defendo a decisão: Transformar em `NaN`; `0`, *nesse caso*, é dado vazio

````python
# DECISÃO: idade 0 significa "não informado". Trocamos por NaN para não
# contaminar as médias. Consequência: as análises de idade usam n=37, não 40.
df["idade"] = df["idade"].astype(float).replace(0, np.nan)

print("média agora:", round(df["idade"].mean(), 2))
print("n de idade: ", int(df["idade"].count()), "de", len(df))
df["idade"].describe().round(2)
````

>⚠️ **ATENÇÃO**: o comentário não é enfeite, é registro de decisão

## 3.3 E se escolhessemos outra opção?
### Ferramentas

````python
print("dropna: descarta as linhas ->", len(df.dropna(subset=["idade"])), "linhas")
print("fillna: preenche com um valor")
print("  com média:  ", round(df["idade"].fillna(df["idade"].mean()).mean(), 2))
print("  com mediana:", round(df["idade"].fillna(df["idade"].median()).mean(), 2))
````

````python
# 🎤 ⚠️ Mas repare no que preencher faz com a dispersão
col = df["idade"]

comparativo = pd.DataFrame({
    "original":    [col.mean(), col.median(), col.std()],
    "com média":   [col.fillna(col.mean()).mean(),   col.fillna(col.mean()).median(),   col.fillna(col.mean()).std()],
    "com mediana": [col.fillna(col.median()).mean(), col.fillna(col.median()).median(), col.fillna(col.median()).std()],
}, index=["média", "mediana", "desvio padrão"]).round(2)

comparativo
````

🗣️ *"Preencher faltante com um valor fixo **sempre** reduz a dispersão (desvio padrão) — vocês
inventaram três pessoas exatamente na média, que não variam nada.*

Quando usar:

Média: poucos valores, com pouca variancia

Mediana: tratar outliers, distribuição torta

NaN: muitos valores vazios; > 20%
