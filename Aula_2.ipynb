### Como fazemos pra melhorar esse código?

````python
nota = 11

if nota > 10:
    print("Nota impossivel")
elif nota >= 9:
    print("Brabissimo")
elif nota >= 7:
    print("A provado")
elif nota >= 5:
    print("Quase lá")
else:
    print("Se lascou")
````

## B.3 for — repetir sem repetir código

O momento mais importante da noite.

````python
notas = [8.5, 5.0, 9.2, 4.1, 7.0]


for nota in notas:
    print(nota)
````

````python
frutas = ["Banana", "Maça", "Limão"]

for fruta in frutas:
  print(f"Eu adoro {fruta}")
````


### Qual é a média dessas notas?



````python
notas = [8.5, 5.0, 9.2, 4.1, 7.0]

# PSEDOCODIGO
# 1. Para cada um dos elementos, somar ao meu total
# 2. Dividir o meu total pelo número de notas

total = 0

for nota in notas:
  total += nota

total / len(notas)
````
