## Aquele recap

````python
notas = [8.5, 5.0, 9.2, 4.1, 7.0]
soma = 0

for nota in notas:
    soma = soma + nota

print("Soma:", soma)
print("Média:", soma / len(notas))
````

- `notas` é uma lista com 5 elementos, todos float
- `soma` é o nosso acumulador
- itera sobra a list com `for`:
  - `nota` é o placeholder: um nome arbitrário que carrega o valor de cada um dos elementos
  - o bloco dentro do `for` vai ser aplicado a todos os elementos

## 2. Lista de dicionários


````python
turma = [
    {"nome": "Ana",   "idade": 28, "cidade": "São Paulo",      "nota": 8.5},
    {"nome": "Bruno", "idade": 35, "cidade": "Rio de Janeiro", "nota": 5.0},
    {"nome": "Carla", "idade": 22, "cidade": "São Paulo",      "nota": 9.2},
    {"nome": "Diego", "idade": 41, "cidade": "Salvador",       "nota": 4.1},
    {"nome": "Eva",   "idade": 30, "cidade": "São Paulo",      "nota": 7.0},
]
````

### Percorrendo a tabela

````python
for aluno in turma:
  print(f"Meu nome é {aluno["nome"]}, moro em {aluno["cidade"]}, tenho {aluno["idade"]} anos.")
````

### Quem passou?

````python
MEDIA = 7.0

for aluno in turma:
  nota = int(aluno["nota"])
  nome = aluno["nome"]

  if nota >= MEDIA:
    print(f"Eu sou {nome} e passei com {nota}")
  else:
    print(f"Eu sou {nome} e reprovei com {nota}")
````

### Qual a média das notas?

````python
soma = 0

for aluno in turma:
  soma += aluno["nota"]

media = soma / len(turma)

print("Soma:", soma)
print("Média:", media)
````

## Extra 4: List comprehension

>Um atalho

>Um jeito diferente de escrever um loop com `for`

>Vamos dobrar os valores dessa lista de `numeros`

````python
numeros = [1, 2, 3, 4, 5]
````

````python
# O jeito que vocês já conhecem:
dobros = []
for n in numeros:
    dobros.append(n * 2)
print(dobros)
````

````python
# List Comprehension
dobro = [n * 2 for n in numeros]
print(dobros)
````

#### Sintaxe:
```
[<código a ser retornado a cada linha> for <placeholder> in <list>]
```

> nota: `[]` são obrigatórios


````python
# ➕ Sobre a tabela: extraindo uma coluna
turma = [
    {"nome": "Ana",   "nota": 8.5},
    {"nome": "Bruno", "nota": 5.0},
    {"nome": "Carla", "nota": 9.2},
]

nomes = [aluno["nome"] for aluno in turma]
notas = [aluno["nota"] for aluno in turma]

print(nomes)
print(notas)
print("Média:", sum(notas) / len(notas))
````

## Extra 5: Capturando erro ao invés de quebrar

````python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Não dá para dividir por zero.")
        return None

print(divide(10, 2))
print(divide(10, 0))
print("o programa continuou rodando")
````

>Sem o `try`, `print(divide(10, 0))` quebraria o código com `ZeroDivisionError`

>`except ZeroDivisionError` diz para o programa o que fazer se esse erro acontecer

## Extra 6: `while`

>`while` recebe uma condição e continua rodando o bloco de código até que essa condição seja falsa

````python
saldo = 100
mes = 0

while saldo > 0:
    mes = mes + 1
    print("mês", mes, "| saldo:", saldo)
    saldo = saldo - 30

print("O dinheiro acabou no mês", mes)
````

### E se criarmos nossa própria classe?

````python
class Aluno:
    def __init__(self, nome, notas):
        self.nome = nome          # atributo: o que ele TEM
        self.notas = notas

    def media(self):              # método: o que ele FAZ
        return sum(self.notas) / len(self.notas)


ana = Aluno("Ana", [8.0, 7.5, 9.0])
print(ana.nome)          # atributo: SEM parênteses
print(ana.media())       # método: COM parênteses
````

- `__init__` é o que roda quando fazemos `Aluno()`
- `Aluno("Ana", [8.0, 7.5, 9.0])` passa `nome` e `notas` mas não passa `self`
- `self.nome` dentro de `__init__` cria atributo `nome` para cada instância de `Aluno`
- `def media(self)` cria uma função de instância. Você pode chamar `Aluno("Ana", [8.0, 7.5, 9.0]).media()`
