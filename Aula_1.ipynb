```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
rng = np.random.default_rng(7)

turma = pd.DataFrame({
    "idade": rng.integers(18, 58, 300),
    "renda": np.round(rng.lognormal(7.6, 0.55, 300), 2),
    "estado": rng.choice(["SP", "RJ", "BA", "MG"], 300, p=[.5, .2, .2, .1]),
})

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data=turma, x="idade", bins=20, ax=ax[0]).set_title("Quem são os alunos")
sns.boxplot(data=turma, x="estado", y="renda", ax=ax[1]).set_title("Renda por estado")
plt.tight_layout(); plt.show()
```

## 10. Métodos de string

`.strip()` - Se livra do espaço antes e depois

`.upper()` - DEIXA TODAS AS LETRAS MAIÚSCULAS

`.lower()` - deixa todas as letras minúsculas

`.replace(a, b)` - Troca a parte da string `a` por `b`

`.split(sep)` - Retorna uma lista de strings menores, separadas por `sep`

`len(texto)` - Retorna o número de caractéres da string

```python
frase = "  Toti Diversidade  "

frase.strip()

frase.upper()
frase.lower()

frase.strip().replace("Diversidade", "Tecnologia")

frase.strip().split(" ")

len(frase.strip())
```
